## Q6 · Median value of horsepower

**Question.** Does mode-based imputation change the median of `horsepower`?

</div>

| Step | Operation |
| :---: | :--- |
| 1 | Calculate the original median. |
| 2 | Find the most frequent value. |
| 3 | Fill missing values with that mode. |
| 4 | Calculate the median again. |

| Option | Outcome |
| :---: | :--- |
| A | Yes, it increased |
| B | Yes, it decreased |
| C | No |

```python
# Calculate the median before imputation
median_before = df["horsepower"].median()

# Find the most frequent horsepower value
mode_value = df["horsepower"].mode()[0]

# Fill missing values without changing the original DataFrame
horsepower_filled = df["horsepower"].fillna(mode_value)

# Calculate the median after imputation
median_after = horsepower_filled.median()

pd.Series(
    {
        "Median before imputation": median_before,
        "Mode used for imputation": mode_value,
        "Median after imputation": median_after,
    }
)
```

<div style="padding: 14px 18px; border-left: 5px solid #16a34a; border-radius: 8px; background: #f0fdf4;">

**Result · Q6**
**`Yes, it decreased`** — from **`254.0`** to **`252.0`**.

<small>Imputing 877 missing values with the mode (252.0) added mass below the original median.</small>

</div>

## Q7 · Sum of regression weights

**Question.** What is the sum of the weights returned by the normal equation?

</div>

| Step | Linear algebra operation |
| :---: | :--- |
| 1 | Select the first 7 cars from Asia. |
| 2 | Build `X` from `vehicle_weight` and `model_year`. |
| 3 | Compute `XTX = X.T @ X` and invert it. |
| 4 | Calculate `w = inv(XTX) @ X.T @ y`. |
| 5 | Return `w.sum()`. |

| Option | Sum of weights |
| :---: | ---: |
| A | 0.0369 |
| B | 0.369 |
| C | 3.69 |
| D | 36.9 |

```python
import numpy as np

# Select the first seven cars from Asia and the required features
X = df.loc[df["origin"] == "Asia", ["vehicle_weight", "model_year"]].iloc[:7].to_numpy()

# Calculate the normal equation components
XTX = X.T @ X
inv_XTX = np.linalg.inv(XTX)

# Define the target vector and calculate the weights
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = inv_XTX @ X.T @ y

w.sum()
```

<div style="padding: 14px 18px; border-left: 5px solid #16a34a; border-radius: 8px; background: #f0fdf4;">

**Result · Q7**
The sum of all elements in vector $w$ is approximately **`0.369`** (`0.3691969...`).
